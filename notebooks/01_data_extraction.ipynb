```python
import os
import yfinance as yf
import pandas as pd

tickers = {
    "MSFT":   "MSFT",
    "IBM":    "IBM",
    "GOOGL":  "GOOGL",
    "USDJPY": "JPY=X",
    "GBPUSD": "GBPUSD=X",
    "SP500":  "^GSPC",
    "DOW":    "^DJI",
    "VIX":    "^VIX",
}

start = "2010-01-01"
end = pd.Timestamp.today().strftime("%Y-%m-%d")

raw = yf.download(list(tickers.values()), start=start, end=end, auto_adjust=True)

close = raw["Close"].rename(columns={v: k for k, v in tickers.items()})
close = close[list(tickers.keys())]

print(close.shape)
print(close.isna().sum())
print(close.head())
print(close.tail())

os.makedirs("../data", exist_ok=True)
close.to_csv("../data/raw_close_prices.csv")
```
